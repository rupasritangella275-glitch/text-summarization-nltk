```python
import numpy as np
import soundfile as sf

sample_rate = 16000
duration = 3  # seconds

audio = np.zeros(int(sample_rate * duration), dtype=np.float32)
sf.write("sample_audio.wav", audio, sample_rate)

print("✅ Dummy audio file created: sample_audio.wav")
```

```python
import speech_recognition as sr
import os

def speech_to_text(audio_file):
    if not os.path.exists(audio_file):
        print("❌ File not found:", audio_file)
        return

    r = sr.Recognizer()
    with sr.AudioFile(audio_file) as source:
        audio = r.record(source)

    try:
        text = r.recognize_google(audio)
        print("📝 Recognized text:", text)
    except sr.UnknownValueError:
        print("⚠️ No speech detected (expected for dummy audio).")
    except sr.RequestError as e:
        print("❌ API error:", e)

speech_to_text("sample_audio.wav")
```
