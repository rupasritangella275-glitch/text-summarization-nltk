```python

import os
import warnings
from transformers import pipeline, logging

# ---------------- CLEANUP (remove warnings) ----------------
warnings.filterwarnings("ignore")
logging.set_verbosity_error()
os.environ["TRANSFORMERS_NO_ADVISORY_WARNINGS"] = "1"

# ---------------- LOAD MODEL ----------------
generator = pipeline(
    "text-generation",
    model="gpt2",
    device=-1  # CPU (Colab default)
)

# ---------------- INPUT PROMPT ----------------
prompt = "Artificial Intelligence is changing the world because"

# ---------------- GENERATE TEXT ----------------
output = generator(
    prompt,
    max_new_tokens=50,          # use this instead of max_length
    do_sample=True,             # better text
    temperature=0.7,            # controls creativity
    top_p=0.9,                  # nucleus sampling
    pad_token_id=50256          # suppress padding warning
)

# ---------------- OUTPUT ----------------
print("Input Prompt:")
print(prompt)

print("\nGenerated Text:")
print(output[0]["generated_text"])
```
